## 0. Jetson 초기 설정

### A. 한글 설치

```bash
sudo apt install ibus ibus-hangul -y
```

설치 후 재부팅:

```bash
reboot
```

### B. 프로젝트 폴더 생성

```bash
mkdir vision-llm
```

### C. Visual Studio Code 설치

```bash
wget -O vscode.deb "https://update.code.visualstudio.com/latest/linux-deb-arm64/stable"
sudo apt install ./vscode.deb
```

### D. Visual Studio Code Extensions 설치

1. Visual Studio Code 실행
2. 좌측 메뉴의 **Extensions** 선택
3. ***Python***과 ***Jupyter*** 설치

### E. Visual Studio Code 가상 환경 생성

1. Visual Studio Code에서 `Ctrl` + `Shift` + `~` 눌러 터미널 열기

2. `python3-venv` 패키지 설치:
```bash
sudo apt install python3-venv
```

3. 가상 환경 생성:
```bash
python3 -m venv --system-site-packages .venv
```

4. 가상 환경 활성화:
```bash
source .venv/bin/activate
```

5. `pip` 업그레이드:
```bash
python -m pip install --upgrade pip
```

6. `jupyter`, `ipykernel` 설치
```bash
pip install jupyter ipykernel
```

7. Jupyter 커널 목록에 등록
```bash
python -m ipykernel install --user --name=vision_llm --display-name "Vision-LLM (Python3.10)"
```

8. VS Code 닫기 & 재실행

9. 우측 상단에 `Select Kernel` → `Jupyter Kernel...` → `Vision-LLM (Python3.10)` 선택

### F. Chromium 삭제

Chromium을 삭제하지 않으면 실습 중 Jetson이 정상적으로 작동하지 않을 수 있습니다.

필수로 삭제해줍니다.

```bash
sudo snap remove --purge chromium
sudo apt purge chromium-browser chromium -y
flatpak uninstall org.chromium.Chromium -y
```
